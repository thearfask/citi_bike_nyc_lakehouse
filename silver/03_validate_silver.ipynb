```python
from pyspark.sql import functions as F
from pyspark.sql import Window
```

```python
trips_df = spark.read.table("nyc_citi_bike_catalog.bronze.raw_trips_tbl")
```

```python
trips_df.display()
```

```python
valid_trips_df = trips_df.filter(F.col("ride_id").isNotNull())
invalid_trips_df = trips_df.filter(F.col("ride_id").isNull())
```

```python
valid_trips_df.count(), invalid_trips_df.count()
```

```python
win_spec = Window.partitionBy(['ride_id', 'source_key']).orderBy(F.col('started_at'))
valid_trips_df1 = valid_trips_df.withColumn("row_num", F.row_number().over(win_spec))
```

```python
valid_trips_df1.filter(F.col('row_num')> 1).display()
```

```python
valid_trips_df1.filter(F.col('ride_id') == "FF41BAE64F443E05").display()
```

```python
from pyspark.sql.types import StructType, StructField, StringType, TimestampType, FloatType

trips_schema = StructType([
    StructField("ride_id", StringType(), True),
    StructField("rideable_type", StringType(), True),
    StructField("started_at", TimeStampType(), True),
    StructField("ended_at", TimeStampType(), True),
    StructField("start_station_name", StringType(), True),
    StructField("start_station_id", StringType(), True),
    StructField("end_station_name", StringType(), True),
    StructField("end_station_id", StringType(), True),
    StructField("start_lat", FloatType(), True),
    StructField("start_lng", FloatType(), True),
    StructField("end_lat", FloatType(), True),
    StructField("end_lng", FloatType(), True),
    StructField("member_casual", StringType(), True),
    StructField("source_key", StringType(), True),
    StructField("batch_id", StringType(), True)
])

print(trips_schema)
```
