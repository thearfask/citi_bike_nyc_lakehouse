```python
from pyspark.sql import functions as F
```

```python
trips_df = spark.read.table("nyc_citi_bike_catalog.bronze.raw_trips_tbl")
```

```python
trips_df.display()
```

```python
valid_trips_df = trips_df.filter(F.col("ride_id").isNotNull())
invalid_trips_df = trips_df.filter(F.col("ride_id").isNull())
```

```python
valid_trips_df.count(), invalid_trips_df.count()
```
